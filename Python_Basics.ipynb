# Python Basics
A practical introduction to the core concepts of Python with simple explanations and examples.

## 1. Variables and Data Types
Variables store values. Python automatically determines the data type based on the assigned value.

```python
name = 'Samaher'
age = 25
pi = 3.14
is_learning = True
print(name, age, pi, is_learning)
```

```python
print(type(name))
print(type(age))
print(type(pi))
print(type(is_learning))
```

## 2. Strings
Strings represent text. They can be indexed, sliced, combined, and modified using built-in methods.

```python
text = 'Python Programming'
print(text.upper())
print(text[:6])
print(len(text))
```

## 3. Lists
A list is an ordered and mutable collection. It can store multiple values and different data types.

```python
skills = ['Python', 'SQL', 'Machine Learning']
skills.append('Pandas')
print(skills)
print(skills[0])
```

## 4. Tuples, Sets, and Dictionaries
A tuple is ordered and immutable. A set stores unique values. A dictionary stores key-value pairs.

```python
coordinates = (10, 20, 30)
tools = {'Python', 'SQL', 'Python'}
student = {'name': 'Samaher', 'field': 'Data Science'}
print(coordinates)
print(tools)
print(student['field'])
```

## 5. Conditional Statements
Conditions allow a program to make decisions using `if`, `elif`, and `else`.

```python
score = 92
if score >= 90:
    print('Excellent')
elif score >= 70:
    print('Good')
else:
    print('Keep practicing')
```

## 6. Loops
Loops repeat a block of code. A `for` loop iterates over a sequence, while a `while` loop repeats while a condition is true.

```python
for number in range(1, 6):
    print(number)
```

```python
count = 1
while count <= 3:
    print(count)
    count += 1
```

## 7. Functions
Functions group reusable code. Parameters provide input and `return` sends a result back.

```python
def square(number):
    return number ** 2

print(square(5))
```

## 8. List Comprehensions
List comprehensions provide a concise way to create lists from existing iterables.

```python
squares = [number ** 2 for number in range(1, 6)]
print(squares)
```

## 9. Exception Handling
`try` and `except` help handle runtime errors without stopping the entire program.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print('Cannot divide by zero.')
```

## 10. NumPy Basics
NumPy is commonly used for numerical computing and efficient array operations.

```python
import numpy as np
numbers = np.array([1, 2, 3, 4, 5])
print(numbers)
print(numbers.mean())
print(numbers.sum())
```

## 11. pandas Basics
pandas is used to work with structured and tabular data using objects such as DataFrames.

```python
import pandas as pd
data = {'Name': ['Samaher', 'Sara', 'Nora'], 'Score': [95, 88, 91]}
df = pd.DataFrame(data)
df
```

## 12. Small Practice Example
This example combines a list, loop, condition, and function to classify scores.

```python
def check_result(score):
    if score >= 60:
        return 'Pass'
    return 'Fail'

scores = [95, 62, 48, 81]
for score in scores:
    print(score, '->', check_result(score))
```

## Key Takeaways
- Python uses simple and readable syntax.
- Collections such as lists and dictionaries organize data.
- Conditions and loops control program flow.
- Functions make code reusable.
- NumPy and pandas provide useful foundations for data analysis.
